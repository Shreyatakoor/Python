```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
data=fetch_california_housing(as_frame=True)
housing_df=data.frame
numerical_features=housing_df.select_dtypes(include=[np.number])
numerical_features.head()
plt.figure(figsize=(15,10))
for i,feature in enumerate(numerical_features.columns):
    plt.subplot(3,3,i+1)
    sns.histplot(data=housing_df[feature],kde=True,bins=45,color='blue')
    plt.title(f'Histogram of {feature}')
plt.tight_layout()
plt.show()
plt.figure(figsize=(15,10))
for i,feature in enumerate(numerical_features.columns):
    plt.subplot(3,3,i+1)
    sns.boxplot(data=housing_df[feature],color='orange')
    plt.title(f'Box Plot of {feature}')
plt.tight_layout()
plt.show()
print("Description of Outliers")
outliers_summary={}
for feature in numerical_features.columns:
        Q1=housing_df[feature].quantile(0.25)
        Q3=housing_df[feature].quantile(0.75)
        IQR=Q3-Q1
        lower_bound=Q1-1.5*IQR
        upper_bound=Q3+1.5*IQR
        Outliers=housing_df[(housing_df[feature]<lower_bound)|(housing_df[feature]>upper_bound)]
        outliers_summary[feature]=len(Outliers)
print(outliers_summary)
print(housing_df.describe())


```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

# Load Dataset
data = fetch_california_housing(as_frame=True)
housing_df = data.frame

# Select numeric features
numerical_features = housing_df.select_dtypes(include=[np.number])

# ---------------------------------------------------
# 1️⃣ Line Plot
plt.figure(figsize=(10,5))
housing_df['MedHouseVal'].plot(kind='line')
plt.title("Line Plot - Median House Value")
plt.xlabel("Index")
plt.ylabel("MedHouseVal")
plt.show()

# ---------------------------------------------------
# 2️⃣ Scatter Plot
plt.figure(figsize=(8,6))
housing_df.plot(kind='scatter', x='MedInc', y='MedHouseVal')
plt.title("Scatter Plot - MedInc vs MedHouseVal")
plt.show()

# ---------------------------------------------------
# 3️⃣ Bar Plot (Mean of Features)
plt.figure(figsize=(12,6))
housing_df.mean().plot(kind='bar')
plt.title("Bar Plot - Mean of Features")
plt.xticks(rotation=45)
plt.show()

# ---------------------------------------------------
# 4️⃣ Pie Chart (HouseAge Top 10)
plt.figure(figsize=(6,6))
housing_df['HouseAge'].head(10).value_counts().plot(
    kind='pie', autopct='%1.1f%%'
)
plt.title("Pie Chart - HouseAge (Top 10)")
plt.ylabel("")
plt.show()

# ---------------------------------------------------
# 5️⃣ Area Plot
plt.figure(figsize=(10,6))
housing_df[['MedInc','MedHouseVal']].head(100).plot(
    kind='area', alpha=0.5
)
plt.title("Area Plot")
plt.show()

# ---------------------------------------------------
# 6️⃣ Density Plot
plt.figure(figsize=(8,5))
housing_df['MedInc'].plot(kind='density')
plt.title("Density Plot - MedInc")
plt.show()

# ---------------------------------------------------
# 7️⃣ Hexbin Plot
plt.figure(figsize=(8,6))
housing_df.plot(kind='hexbin',
                x='MedInc',
                y='MedHouseVal',
                gridsize=25,
                cmap='Blues')
plt.title("Hexbin Plot")
plt.show()

# ---------------------------------------------------
# 8️⃣ Correlation Heatmap
corr = housing_df.corr()
plt.figure(figsize=(10,8))
plt.imshow(corr, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=90)
plt.yticks(range(len(corr)), corr.columns)
plt.title("Correlation Heatmap")
plt.show()

# ---------------------------------------------------
# 9️⃣ Violin Plot
plt.figure(figsize=(8,6))
plt.violinplot(housing_df['MedInc'])
plt.title("Violin Plot - MedInc")
plt.show()

# ---------------------------------------------------
# 🔟 Stem Plot
plt.figure(figsize=(8,5))
plt.stem(housing_df['MedInc'].head(30))
plt.title("Stem Plot - First 30 MedInc")
plt.show()

# ---------------------------------------------------
# 1️⃣1️⃣ Histogram (All Numeric Features)
plt.figure(figsize=(15,10))
for i, feature in enumerate(numerical_features.columns):
    plt.subplot(3,3,i+1)
    plt.hist(housing_df[feature], bins=30)
    plt.title(f'Histogram of {feature}')
plt.tight_layout()
plt.show()

# ---------------------------------------------------
# 1️⃣2️⃣ Box Plot (All Numeric Features)
plt.figure(figsize=(15,10))
for i, feature in enumerate(numerical_features.columns):
    plt.subplot(3,3,i+1)
    plt.boxplot(housing_df[feature])
    plt.title(f'Box Plot of {feature}')
plt.tight_layout()
plt.show()

print("All Plots Generated Successfully ✅")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# Load Iris dataset
iris = load_iris()
data = iris.data
target = iris.target
label_names = iris.target_names

# Create DataFrame
iris_df = pd.DataFrame(data, columns=iris.feature_names)
iris_df['target'] = target

# Apply PCA
pca = PCA(n_components=2)
data_reduced = pca.fit_transform(data)

reduced_df = pd.DataFrame(data_reduced, columns=['PC1', 'PC2'])
reduced_df['target'] = target

# ---------------------------------------------------
# 1. Line Plot
plt.figure(figsize=(8,5))
plt.plot(iris_df.iloc[:50,0])
plt.title("Line Plot")
plt.show()

# ---------------------------------------------------
# 2. Scatter Plot (PCA)
plt.figure(figsize=(8,6))
colors = ['red','green','blue']

for i in range(3):
    plt.scatter(
        reduced_df[reduced_df['target']==i]['PC1'],
        reduced_df[reduced_df['target']==i]['PC2'],
        c=colors[i],
        label=label_names[i]
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Scatter Plot")
plt.legend()
plt.show()

# ---------------------------------------------------
# 3. Bar Plot
plt.figure(figsize=(8,5))
iris_df.iloc[:,:4].mean().plot(kind='bar')
plt.title("Bar Plot")
plt.show()

# ---------------------------------------------------
# 4. Horizontal Bar Plot
plt.figure(figsize=(8,5))
iris_df.iloc[:,:4].mean().plot(kind='barh')
plt.title("Horizontal Bar Plot")
plt.show()

# ---------------------------------------------------
# 5. Histogram
plt.figure(figsize=(8,5))
plt.hist(iris_df['sepal length (cm)'], bins=20)
plt.title("Histogram")
plt.show()

# ---------------------------------------------------
# 6. Box Plot
plt.figure(figsize=(8,5))
plt.boxplot(iris_df.iloc[:,:4])
plt.title("Box Plot")
plt.show()

# ---------------------------------------------------
# 7. Violin Plot
plt.figure(figsize=(8,5))
plt.violinplot(iris_df.iloc[:,:4].values)
plt.title("Violin Plot")
plt.show()

# ---------------------------------------------------
# 8. Pie Chart
plt.figure(figsize=(6,6))
counts = iris_df['target'].value_counts()
plt.pie(counts, labels=label_names, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()

# ---------------------------------------------------
# 9. Area Plot
iris_df.iloc[:20,:4].plot(kind='area', figsize=(8,5))
plt.title("Area Plot")
plt.show()

# ---------------------------------------------------
# 10. Stem Plot
plt.figure(figsize=(8,5))
plt.stem(iris_df['sepal length (cm)'].head(20))
plt.title("Stem Plot")
plt.show()

# ---------------------------------------------------
# 11. Heatmap using Matplotlib
corr = iris_df.iloc[:,:4].corr()

plt.figure(figsize=(8,6))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=45)
plt.yticks(range(len(corr)), corr.columns)
plt.title("Correlation Heatmap")
plt.show()

# ---------------------------------------------------
# 12. Subplots
iris_df.iloc[:,:4].plot(subplots=True, layout=(2,2), figsize=(10,8))
plt.suptitle("Subplots")
plt.show()

print("All plot types generated successfully.")
```
